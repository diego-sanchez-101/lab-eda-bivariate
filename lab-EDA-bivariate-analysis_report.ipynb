Business Report: Best-Seller Category Analysis

Initial Observation At first glance, products in the Women category have far more best sellers, followed by Women's Sport & Outdoor Shoes.

Statistical Significance If p-value is below 0.05, there is a significant association between the two variables. The analysis confirms that category matters for a best seller — the category a product belongs to does affect how likely it is to be a best-seller.

Top Categories by Proportion When considering proportions (best-sellers out of total, with false included), the picture changes. Grocery, Smart Home, and Health & Personal Care complete the top three.
